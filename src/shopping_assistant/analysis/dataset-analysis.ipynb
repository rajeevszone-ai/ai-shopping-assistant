```python
import os

from dotenv import load_dotenv

load_dotenv()
token = os.environ["HF_TOKEN"]
```

Load meta cagetories dataset for adevnture sports

```python
from datasets import load_dataset

print("Streaming sample data from Hugging Face...")

# Stream the metadata JSONL file without using the deprecated dataset script.
meta_dataset = load_dataset(
    "json",
    data_files={
        "full": (
            "hf://datasets/McAuley-Lab/Amazon-Reviews-2023/"
            "raw/meta_categories/meta_Sports_and_Outdoors.jsonl"
        )
    },
    split="full",
    streaming=True,
    token=token,
)
```

```python
import pandas as pd

# Convert a small streamed sample to a DataFrame for inspection.
sample_meta_df = pd.DataFrame(list(meta_dataset.take(1000)))
sample_meta_df.head(5)
```
